# Automated PDF Map Package Builder

# Key Features:
- Automatically merges multiple PDF documents into one.
- Adds bold page numbers to each page, right-aligned at the bottom.
- Maintains original layout and formatting of source PDFs.
- Flexible and reusable — works with any number of PDFs!

## Workflow (Step-by-Step):

### Import Libraries
- Note: Make sure to download and install these libraries before you run the script. 

```python
import os
from PyPDF2 import PdfFileReader, PdfFileWriter
from reportlab.pdfgen import canvas
from reportlab.lib.pagesizes import letter
from reportlab.lib.units import inch
from reportlab.lib import colors
from io import BytesIO
```

### Define the input and output PDF paths:

```python
# Paths
pdf1 = r"C:\Users\dshrestha\ ... \Report Summary1.pdf"
pdf2 = r"C:\Users\dshrestha\... \ Report Summary2.pdf"
#pdf3 = r"C:\Users\dshrestha\... \ Report Summary3.pdf"

output_pdf = r"C:\Users\dshrestha\...\Final Report Summary.pdf"
```

### Read and extract all pages from each PDF using PyPDF2.
#### For each page:
 - Generate a canvas overlay with the page number using reportlab.
 - Merge this overlay into the page as a footer.
 - Combine all modified pages into one output file.

```python
# Read PDFs
reader1 = PdfFileReader(open(pdf1, "rb"))
reader2 = PdfFileReader(open(pdf2, "rb"))
#reader3 = PdfFileReader(open(pdf3, "rb"))
writer = PdfFileWriter()

# Combine pages
all_pages = [reader1.getPage(i) for i in range(reader1.getNumPages())] + \
            [reader2.getPage(i) for i in range(reader2.getNumPages())] #+ \
           # [reader3.getPage(i) for i in range(reader3.getNumPages())]

# Add bold page numbers
for i, page in enumerate(all_pages, start=1):
    packet = BytesIO()
    can = canvas.Canvas(packet, pagesize=letter)
    
    # Set bold font for page number
    can.setFont("Helvetica-Bold", 16)
    can.drawRightString(16.5 * inch, 0.3 * inch, str(i))
    can.save()
    
    packet.seek(0)

    number_pdf = PdfFileReader(packet)
    page.mergePage(number_pdf.getPage(0))
    writer.addPage(page)
```

### Save the final result as:

```python
# Save the result
with open(output_pdf, "wb") as out:
    writer.write(out)

print(f"Final Report: {output_pdf}")
```
