# Unzip Multiple Folders

# 🔧 Key Features: 
- ✅ Unzips all GIS data files in bulk
- ✅ Automatically renames folders with "Fort Worth_" prefix
- ✅ Keeps your datasets organized and ready for analysis
- ✅ Simple, reusable, and saves hours of manual effort

```python
import os
import zipfile

# ✅ Define source and destination folders
source_folder = r"C:\Users\dshrestha\Downloads\FortWorth-GIS-Data\Zipped_GISData"
destination_folder = r"C:\Users\dshrestha\Downloads\FortWorth-GIS-Data\UnZipped_GISData"

# ✅ Ensure destination folder exists
os.makedirs(destination_folder, exist_ok=True)

try:
    for file_name in os.listdir(source_folder):
        if file_name.lower().endswith('.zip'):
            zip_path = os.path.join(source_folder, file_name)
            extract_folder = os.path.join(destination_folder, os.path.splitext(file_name)[0])
            os.makedirs(extract_folder, exist_ok=True)

            with zipfile.ZipFile(zip_path, 'r') as zip_ref:
                zip_ref.extractall(extract_folder)
                print(f"✅ Extracted: {file_name} to {extract_folder}")
except OSError as e:
    print(f"❌ Error accessing directory: {e}")
except Exception as e:
    print(f"❌ An error occurred: {e}")
```

```python
import os
import zipfile

# ✅ Define source and destination folders
source_folder = r"C:\Users\dshrestha\Downloads\FortWorth-GIS-Data\Zipped_GISData"
destination_folder = r"C:\Users\dshrestha\Downloads\FortWorth-GIS-Data\UnZipped_FortWorth_GISData"

# ✅ Ensure destination folder exists
os.makedirs(destination_folder, exist_ok=True)

try:
    for file_name in os.listdir(source_folder):
        if file_name.lower().endswith('.zip'):
            zip_path = os.path.join(source_folder, file_name)

            # 👇 Add "Fort Worth" prefix to folder name
            base_name = os.path.splitext(file_name)[0]
            extract_folder_name = f"Fort Worth_{base_name}"
            extract_folder = os.path.join(destination_folder, extract_folder_name)

            os.makedirs(extract_folder, exist_ok=True)

            with zipfile.ZipFile(zip_path, 'r') as zip_ref:
                zip_ref.extractall(extract_folder)
                print(f"✅ Extracted: {file_name} to {extract_folder}")
except OSError as e:
    print(f"❌ Error accessing directory: {e}")
except Exception as e:
    print(f"❌ An error occurred: {e}")
```
