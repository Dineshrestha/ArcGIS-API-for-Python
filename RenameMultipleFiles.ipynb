Renaming files might be tedious when you have hundreds of files in a folder and you want to make few changes on those files.

Here is a script that rename all files in folder based on a file of a choice. In this script, all it does it replace and reposition "Date" portion from file names.

Example:
You need a file in this format: "LCMAP_2022July01_Percent.img" but you accidently saved all files in this format: "LCMAP_July012022_Percent.img".  The script below helps you rename all files at once.

```python
import os
from pathlib import Path
import re

# Specify the folder path where the files are located
folder_path = r"C:\Users\dshrestha\Downloads\LCMAP_July012023_Percent"

# Create a Path object for the folder
folder = Path(folder_path)

# Regular expression pattern to match the date in the file names
pattern = r"(\d{4})(\w+)(\d{2})"

# Iterate over each file in the folder
for file_path in folder.iterdir():
    # Check if the item is a file
    if file_path.is_file():
        # Split the file name and extension
        base_name = file_path.stem
        extension = file_path.suffix

        # Extract the date components from the file name using regular expression
        match = re.search(pattern, base_name)

        # If the pattern does not match, skip the file and continue to the next one
        if match is None:
            continue

        # Extract the date components from the match object
        year, month, day = match.group(1), match.group(2), match.group(3)

        # Rearrange the components to form the new file name
        new_filename = f"{base_name.split('_')[0]}_{month}{day}{year}_{base_name.split('_')[2]}{extension}"

        # Construct the full path for the new file name
        new_file_path = folder.joinpath(new_filename)

        # Rename the file
        file_path.rename(new_file_path)

print("File names have been successfully changed.")
```
