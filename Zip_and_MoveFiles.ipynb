## Move files to folders

Here's a script that moves files to their respective folders based on matching file names:

```python
import os
import shutil

# Specify the path to the folder containing the files and folders
parent_folder = r"C:\Users\dshrestha\Downloads\Test"

# Iterate over each item in the parent folder
for item in os.listdir(parent_folder):
    item_path = os.path.join(parent_folder, item)

    # Check if the item is a file
    if os.path.isfile(item_path):
        # Extract the file name without extension
        file_name = os.path.splitext(item)[0]

        # Check if a folder with the same name exists
        folder_path = os.path.join(parent_folder, file_name)
        if os.path.isdir(folder_path):
            # Move the file to the corresponding folder
            destination_path = os.path.join(folder_path, item)
            shutil.move(item_path, destination_path)
            print(f"File '{item}' has been moved to folder '{file_name}'.")
        else:
            print(f"Folder '{file_name}' does not exist. Skipping file '{item}'.")
    else:
        print(f"Item '{item}' is not a file. Skipping.")

print("Files have been moved to their respective folders.")
```

## Zip all folder

Here's a script that zips all the folders within a specified folder:

```python
import os
import zipfile

# Specify the path to the folder containing the folders to be zipped
parent_folder = r"C:\Users\dshrestha\Downloads\Test"

# Iterate over each item in the parent folder
for item in os.listdir(parent_folder):
    item_path = os.path.join(parent_folder, item)

    # Check if the item is a directory
    if os.path.isdir(item_path):
        # Create a zip file for the current folder
        zip_file_path = os.path.join(parent_folder, f"{item}.zip")
        with zipfile.ZipFile(zip_file_path, "w", zipfile.ZIP_DEFLATED) as zipf:
            # Iterate over each file in the current folder
            for folder_item in os.listdir(item_path):
                folder_item_path = os.path.join(item_path, folder_item)
                # Add the file to the zip file
                zipf.write(folder_item_path, os.path.join(item, folder_item))

        print(f"Folder '{item}' has been zipped.")

print("All folders have been zipped.")
```
